```python
# Смотрим на сигналы запястья
eda_wrist = data['signal']['wrist']['EDA']
bvp_wrist = data['signal']['wrist']['BVP']
temp_wrist = data['signal']['wrist']['TEMP']
acc_wrist = data['signal']['wrist']['ACC']

print("EDA форма:", eda_wrist.shape)
print("BVP форма:", bvp_wrist.shape)
print("TEMP форма:", temp_wrist.shape)
print("ACC форма:", acc_wrist.shape)

print("\nEDA - мин/макс/среднее:", 
      round(eda_wrist.min(),3), "/", 
      round(eda_wrist.max(),3), "/", 
      round(eda_wrist.mean(),3))
print("TEMP - мин/макс/среднее:", 
      round(temp_wrist.min(),3), "/", 
      round(temp_wrist.max(),3), "/", 
      round(temp_wrist.mean(),3))
```

```python
fig, axes = plt.subplots(3, 1, figsize=(14, 10))

# EDA сигнал
axes[0].plot(eda_wrist[:1000], color='blue', linewidth=0.8)
axes[0].set_title('EDA — Электродермальная активность (стресс)')
axes[0].set_ylabel('мкСм')

# BVP сигнал
axes[1].plot(bvp_wrist[:1000], color='red', linewidth=0.8)
axes[1].set_title('BVP — Фотоплетизмография (пульс)')
axes[1].set_ylabel('амплитуда')

# TEMP сигнал
axes[2].plot(temp_wrist[:1000], color='green', linewidth=0.8)
axes[2].set_title('TEMP — Температура кожи')
axes[2].set_ylabel('°C')

plt.tight_layout()
plt.show()
```

```python
subjects = ['S2','S3','S4','S5','S6','S7','S8','S9',
            'S10','S11','S13','S14','S15','S16','S17']
data_path = r'C:\Users\admin\stress-detection-wesad\data\WESAD'

all_data = []
fs_label = 700
fs_eda   = 4
fs_bvp   = 64
fs_acc   = 32

for subject in subjects:
    pkl_path = os.path.join(data_path, subject, f'{subject}.pkl')
    with open(pkl_path, 'rb') as f:
        d = pickle.load(f, encoding='latin1')
    
    labels = d['label']
    eda    = d['signal']['wrist']['EDA'].flatten()
    temp   = d['signal']['wrist']['TEMP'].flatten()
    bvp    = d['signal']['wrist']['BVP'].flatten()
    acc    = d['signal']['wrist']['ACC']

    # Прореживаем метки под частоту EDA (4 Гц)
    labels_eda = labels[::fs_label // fs_eda]
    labels_bvp = labels[::fs_label // fs_bvp]
    labels_acc = labels[::fs_label // fs_acc]

    # Агрегируем BVP и ACC до 4 Гц
    min_bvp = min(len(labels_bvp), len(bvp))
    min_acc = min(len(labels_acc), len(acc))

    bvp_mean = bvp[:min_bvp].reshape(-1, fs_bvp // fs_eda).mean(axis=1)
    acc_mag  = np.sqrt((acc[:min_acc]**2).sum(axis=1))
    acc_mean = acc_mag.reshape(-1, fs_acc // fs_eda).mean(axis=1)

    min_final = min(len(labels_eda), len(eda), len(temp), 
                    len(bvp_mean), len(acc_mean))

    df_subj = pd.DataFrame({
        'EDA':     eda[:min_final],
        'TEMP':    temp[:min_final],
        'BVP':     bvp_mean[:min_final],
        'ACC_mag': acc_mean[:min_final],
        'label':   labels_eda[:min_final],  # ВСЕ метки 0,1,2,3,4,6,7
        'subject': subject
    })

    all_data.append(df_subj)
    print(f"{subject}: {len(df_subj)} записей, метки: {df_subj['label'].unique()}")

df_all = pd.concat(all_data, ignore_index=True)
print(f"\nИтого ВСЕ данные: {len(df_all):,} записей")
print(df_all['label'].value_counts())

# Сохраняем в CSV
csv_path = r'C:\Users\admin\stress-detection-wesad\data\wesad_all.csv'
df_all.to_csv(csv_path, index=False)
print(f"\nСохранено в {csv_path}")
```

```python
# Читаем из CSV - теперь всегда так будем начинать
csv_path = r'C:\Users\admin\stress-detection-wesad\data\wesad_all.csv'
df_all = pd.read_csv(csv_path)

print("Все данные загружены:", df_all.shape)
print("\nВсе метки:")
labels_map = {0: 'Не определено', 1: 'Спокойствие', 2: 'Стресс', 
              3: 'Развлечение', 4: 'Медитация', 5: 'Доп.3',
              6: 'Доп.1', 7: 'Доп.2'}
for label, count in df_all['label'].value_counts().items():
    print(f"  {labels_map.get(label, label)}: {count:,}")

# Для ML задачи: стресс vs спокойствие
df = df_all[df_all['label'].isin([1, 2])].copy()
df['label'] = df['label'].map({1: 0, 2: 1})  # 0=спокойствие, 1=стресс

print(f"\nРабочий датасет (стресс vs спокойствие):")
print(f"  Спокойствие: {(df['label']==0).sum():,}")
print(f"  Стресс:      {(df['label']==1).sum():,}")
print(f"  Итого:       {len(df):,}")
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

features = ['EDA', 'TEMP', 'BVP', 'ACC_mag']
colors_calm = 'green'
colors_stress = 'red'

for idx, feature in enumerate(features):
    ax = axes[idx//2][idx%2]
    df[df['label']==0][feature].hist(ax=ax, alpha=0.6, color=colors_calm, 
                                      bins=50, label='Спокойствие')
    df[df['label']==1][feature].hist(ax=ax, alpha=0.6, color=colors_stress, 
                                      bins=50, label='Стресс')
    ax.set_title(f'Распределение {feature}')
    ax.legend()

plt.suptitle('Сравнение сигналов: Спокойствие vs Стресс', fontsize=14)
plt.tight_layout()
plt.show()

# Корреляционная матрица
print("\nСредние значения по классам:")
print(df.groupby('label')[features].mean().round(3))
```
