```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import warnings
warnings.filterwarnings('ignore')

# Настройка графиков
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['font.size'] = 12

print("Все библиотеки загружены!")
```

```python
import pickle

data_path = r'C:\Users\admin\stress-detection-wesad\data\WESAD\S2\S2.pkl'

with open(data_path, 'rb') as f:
    data = pickle.load(f, encoding='latin1')

print("Ключи датасета:", data.keys())
print("Сигналы:", data['signal'].keys())
```

```python
print("Сигналы груди (chest):", data['signal']['chest'].keys())
print("Сигналы запястья (wrist):", data['signal']['wrist'].keys())
print("\nМетки (уникальные):", np.unique(data['label']))
print("Размер меток:", data['label'].shape)
```

```python
# Визуализация распределения меток
labels_map = {0: 'Не определено', 1: 'Спокойствие', 
              2: 'Стресс', 3: 'Развлечение', 
              4: 'Медитация', 6: 'Доп.1', 7: 'Доп.2'}

unique, counts = np.unique(data['label'], return_counts=True)
label_names = [labels_map[l] for l in unique]

plt.figure(figsize=(10, 5))
bars = plt.bar(label_names, counts, color=['gray','green','red','blue','purple','orange','pink'])
plt.title('Распределение состояний — Участник S2')
plt.xlabel('Состояние')
plt.ylabel('Количество записей')
for bar, count in zip(bars, counts):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 10000, 
             f'{count:,}', ha='center', fontsize=9)
plt.tight_layout()
plt.show()
```

```python
# Смотрим на сигналы запястья
eda_wrist = data['signal']['wrist']['EDA']
bvp_wrist = data['signal']['wrist']['BVP']
temp_wrist = data['signal']['wrist']['TEMP']
acc_wrist = data['signal']['wrist']['ACC']

print("EDA форма:", eda_wrist.shape)
print("BVP форма:", bvp_wrist.shape)
print("TEMP форма:", temp_wrist.shape)
print("ACC форма:", acc_wrist.shape)

print("\nEDA - мин/макс/среднее:", 
      round(eda_wrist.min(),3), "/", 
      round(eda_wrist.max(),3), "/", 
      round(eda_wrist.mean(),3))
print("TEMP - мин/макс/среднее:", 
      round(temp_wrist.min(),3), "/", 
      round(temp_wrist.max(),3), "/", 
      round(temp_wrist.mean(),3))
```

```python
fig, axes = plt.subplots(3, 1, figsize=(14, 10))

# EDA сигнал
axes[0].plot(eda_wrist[:1000], color='blue', linewidth=0.8)
axes[0].set_title('EDA — Электродермальная активность (стресс)')
axes[0].set_ylabel('мкСм')

# BVP сигнал
axes[1].plot(bvp_wrist[:1000], color='red', linewidth=0.8)
axes[1].set_title('BVP — Фотоплетизмография (пульс)')
axes[1].set_ylabel('амплитуда')

# TEMP сигнал
axes[2].plot(temp_wrist[:1000], color='green', linewidth=0.8)
axes[2].set_title('TEMP — Температура кожи')
axes[2].set_ylabel('°C')

plt.tight_layout()
plt.show()
```

```python
subjects = ['S2','S3','S4','S5','S6','S7','S8','S9',
            'S10','S11','S13','S14','S15','S16','S17']
data_path = r'C:\Users\admin\stress-detection-wesad\data\WESAD'

all_data = []
fs_label  = 700
fs_eda    = 4
fs_bvp    = 64
fs_acc_w  = 32
fs_chest  = 700

for subject in subjects:
    pkl_path = os.path.join(data_path, subject, f'{subject}.pkl')
    with open(pkl_path, 'rb') as f:
        d = pickle.load(f, encoding='latin1')
    
    labels = d['label']
    
    # --- ЗАПЯСТЬЕ ---
    eda_w  = d['signal']['wrist']['EDA'].flatten()
    temp_w = d['signal']['wrist']['TEMP'].flatten()
    bvp_w  = d['signal']['wrist']['BVP'].flatten()
    acc_w  = d['signal']['wrist']['ACC']
    
    # --- ГРУДЬ ---
    ecg    = d['signal']['chest']['ECG'].flatten()
    eda_c  = d['signal']['chest']['EDA'].flatten()
    emg    = d['signal']['chest']['EMG'].flatten()
    resp   = d['signal']['chest']['Resp'].flatten()
    temp_c = d['signal']['chest']['Temp'].flatten()
    acc_c  = d['signal']['chest']['ACC']

    # Метки → 4 Гц
    step = fs_label // fs_eda  # 175
    labels_4hz = labels[::step]

    # Запястье → 4 Гц
    bvp_mean  = bvp_w[:(len(bvp_w)//(fs_bvp//fs_eda))*(fs_bvp//fs_eda)].reshape(-1, fs_bvp//fs_eda).mean(axis=1)
    acc_w_mag = np.sqrt((acc_w**2).sum(axis=1))
    acc_w_mean = acc_w_mag[:(len(acc_w_mag)//(fs_acc_w//fs_eda))*(fs_acc_w//fs_eda)].reshape(-1, fs_acc_w//fs_eda).mean(axis=1)

    # Грудь → 4 Гц (шаг = 175)
    chest_step = fs_chest // fs_eda  # 175
    ecg_mean   = ecg[:(len(ecg)//chest_step)*chest_step].reshape(-1, chest_step).mean(axis=1)
    eda_c_mean = eda_c[:(len(eda_c)//chest_step)*chest_step].reshape(-1, chest_step).mean(axis=1)
    emg_mean   = emg[:(len(emg)//chest_step)*chest_step].reshape(-1, chest_step).mean(axis=1)
    resp_mean  = resp[:(len(resp)//chest_step)*chest_step].reshape(-1, chest_step).mean(axis=1)
    temp_c_mean = temp_c[:(len(temp_c)//chest_step)*chest_step].reshape(-1, chest_step).mean(axis=1)
    acc_c_mag  = np.sqrt((acc_c**2).sum(axis=1))
    acc_c_mean = acc_c_mag[:(len(acc_c_mag)//chest_step)*chest_step].reshape(-1, chest_step).mean(axis=1)

    # Минимальная длина по всем сигналам
    min_len = min(len(labels_4hz), len(eda_w), len(temp_w),
                  len(bvp_mean), len(acc_w_mean),
                  len(ecg_mean), len(eda_c_mean), len(emg_mean),
                  len(resp_mean), len(temp_c_mean), len(acc_c_mean))

    df_subj = pd.DataFrame({
        # Запястье
        'EDA_wrist':  eda_w[:min_len],
        'TEMP_wrist': temp_w[:min_len],
        'BVP':        bvp_mean[:min_len],
        'ACC_wrist':  acc_w_mean[:min_len],
        # Грудь
        'ECG':        ecg_mean[:min_len],
        'EDA_chest':  eda_c_mean[:min_len],
        'EMG':        emg_mean[:min_len],
        'Resp':       resp_mean[:min_len],
        'TEMP_chest': temp_c_mean[:min_len],
        'ACC_chest':  acc_c_mean[:min_len],
        # Метки
        'label':   labels_4hz[:min_len],
        'subject': subject
    })

    all_data.append(df_subj)
    print(f"{subject}: {len(df_subj)} записей")

df_all = pd.concat(all_data, ignore_index=True)
print(f"\nИтого: {len(df_all):,} записей")
print(f"Признаков: {len(df_all.columns)-2}")
print(df_all['label'].value_counts())

# Сохраняем
csv_path = r'C:\Users\admin\stress-detection-wesad\data\wesad_all.csv'
df_all.to_csv(csv_path, index=False)
print(f"\nСохранено → {csv_path}")
```

```python
# Рабочий датасет - только стресс и спокойствие
df = df_all[df_all['label'].isin([1, 2])].copy()
df['label'] = df['label'].map({1: 0, 2: 1})

features = ['EDA_wrist', 'TEMP_wrist', 'BVP', 'ACC_wrist',
            'ECG', 'EDA_chest', 'EMG', 'Resp', 'TEMP_chest', 'ACC_chest']

# Корреляционная матрица
plt.figure(figsize=(12, 8))
corr = df[features + ['label']].corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Корреляционная матрица признаков')
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 5, figsize=(18, 8))

for idx, feature in enumerate(features):
    ax = axes[idx//5][idx%5]
    df.boxplot(column=feature, by='label', ax=ax)
    ax.set_title(feature)
    ax.set_xlabel('0=Спокойствие, 1=Стресс')
    ax.set_xticklabels(['Спокойствие', 'Стресс'])

plt.suptitle('Боксплоты признаков по классам', fontsize=14)
plt.tight_layout()
plt.show()
```
